## Análise Exploratória de Dados

A análise exploratória de dados (AED/EDA - *Exploratory Data Analysis*) tem como objetivos principais:
* explorar os dados para descobrir ou identificar aspectos ou padrões de maior interesse
* representar os dados para destacar ou chamar a atenção para aspectos e padrões que podem ou não se confirmar após inferências

Na EDA, já assumimos que foi realizada uma limpeza e um pré processamento dos dados.

Para Tukey (1977), a EDA busca evidências que se confirmam após uma análise confirmatória/inferencial. Isso significa que na EDA, trabalhamos com dados amostrais que desejamos que sejam significativos da população sob a qual faremos a análise inferencial posterior.

### A natureza dos dados

Para efetuar a análise exploratória, precisamos entender a natureza dos dados que estamos tratando, que geralmente serão dados estruturados e retangulares, ou seja, possuem *unidades amostrais* (cada linha/ocorrência/evento) e variáveis (colunas/atributos/features).

Para cada tipo de variável, podemos realizar uma análise. Os tipos de variáveis podem ser:
* **Qualitativas (não-numéricas)**
    * Nominais: sexo, cor da pele, fumante/não fumante, adimplente/inadimplente
    * Ordinais: escolaridade, idade (por faixa), grau de satisfação
        * Algumas variáveis ordinais podem até ser numéricas, mas não são tratadas como.
* **Quantitativas (numéricas)**
    * Discretas: número de peças defeituosas, número de produtos vendidos 
    * Contínuas: peso, idade, pressão sanguínea, valor pago em um produto

No processo de Análise Exploratória, podemos realizar uma *análise descritiva* dos dados, isto é, um conjunto de técnicas que resumem a informação de um conjunto de dados. Para cada tipo de variável, temos uma medida descritiva/resumo mais adequada.

* **Medidas de posição**: resumem a localização dos dados
    * Média: boa propriedade de resumo
    * Mediana: resume bem ao mesmo tempo que não é impactada por valores extremos
    * Moda: valor mais frequente de um conjunto de dados
    * Quantis: caracteriza a distribuição dos dados
* **Medidas de dispersão**: determina a variabilidade dos dados
    * Desvio-padrão
    * Variância
    * Amplitude
    * Coefciente de variação: medida de dispersão relativa
* **Assimetria**: assimetria da distribuição dos dados
* **Curtose**: achatamento da distribuição (altura do pico de uma curva de distribuição de dados em comparação com uma distribuição normal)
* **Medidas de associação**: Covariância, Coefciente de correlação de Pearson, Coefciente de correlação de Spearman

Para entender melhor sobre cada uma dessas medidas, ler [Análise Exploratória de Dados, por Cibele Russo](https://github.com/cibelerusso/FundamentosemCienciasdeDados/blob/main/Notebooks/Fundamentos_em_Ci%C3%AAncias_de_Dados_Aula8.ipynb)

## Pré processamento

Se refere às técnicas aplicadas aos dados brutos para transformá-los em um formato adequado para algorimos de ML. É crucial compreender que:
* A qualidade do modelo não é melhor que a qualidade dos dados que entram nele. (*Garbage in, garbage out*)
* Algoritmos são sensíveis a determinados tipos de problemas. É necessário evaluar o que o modelo que está sendo preparado deve ser sensível para escolher o modelo que mais se adequa à realidade dos dados.
    * O modelo deve se adequar aos dados, não o contrário
* Modelos podem não convergir com dados mal preparados

Algumas etapas:
* Tratamento de missing values
    * Algoritmos não aceitam ou tem sua performance afetada negativamente em modelo com dados ausentes
    * Viés são introduzidos se não tratados adequadamente
* Outliers
    * Valores destoando da média
* Duplicatasdados, mas que podem ser de vário
    * Podem causar overfitting, viés e dificultar a análise de algumas métricas
* Feature Scaling
    * Variáveis em diferentes escalas podem dominar algoritmos baseados em distância ou gradiente.
* Encoding de variáveis categóricas
    * Algoritmos de ML trabalham com números, não com textos ou categorias
    * Precisamos representar essas categorias numericamente sem introduzir relações artificiais

### Missing values

O mecanismo que usamos para tratar os valores ausentes dependem dos diferentes motivos pelos quais os missing values existem. Tomemos um exemplo de uma pesquisa virtual:

Ex 1: No mecanismo aleatório, dados são perdidos por uma falha no sistema que deleta aleatoriamente. Isso não gera um viés.
Ex 2: No mecanismo baseado em características, pessoas mais velhas são menos propensas a responder.
Ex 3: No mecanismo baseado na própria renda, pessoas ricas se recusam a responder por privacidade e pessoas pobres sentem vergonha em responder.

1. MCAR (Missing Completly at Random): a chance de está ausente é a mesma pra todos
2. MAR (Missing at Random): a chance depende de uma variável. Existe um viés e conseguimos corrigir esse viés.
3. MNAR (Missing not at Random): uma parte representativa ou total de uma classe não é representada.

### Outliers

São observações que são significamente diferentes do restante dos dados. Podem ser dos seguintes tipos:
* Statistical Outliers: estatísticamente extremos da distribuição de dados, mas que podem ser válidos.
    * Exemplo: em uma empresa, os funcionários recebem entre R$3.000 e R$15.000, enquanto o CEO recebe R$500.000
    * Mantemos os valores, aliás, eles são reais. Deletá-los implica perda de informação valiosa
    * Transformações, como log e sqrt para reduzir impacto
    * Modelos robustos, como Random Forest e SVM não são muito sensíveis a outliers, então podemos priorizar o uso destes
* Domain-Specific Outliers
    * Valores que violam conhecimento específico do domínio, mesmo que sejam estatisticamente plausíveis (logicamente inconsistentes)
    * Exemplos: idade de 150 anos, desconto de 120%, data de nascimento para 2030, peso de 500kg, temperatura corporal de 50°C, pressão arterial de 300/200...
    * Investigar primeiro e corrigir se existir erro e for identificado
    * Deletar se for um valor comprovadamente impossível
    * Capping em limites lógicos
* Error Outliers
    * Valores incorretos devido a erros no processo de coleta, entrada ou transmissão de dados
    * Resultado de erro humano ou técnico, mas que podem ser estatisticamente normais
    * Exemplos: corrupção de bits, truncamento, encoding, timestamps incorretos, default values são substituídos, duplicação por falha de rede
    * Corrigir se possível para voltar à fonte original ou deletar se não possível
    * Imputação cuidadosa se padrão for identificado
    * Melhoria de processos para previnir reincidência desse tipo de erro

### Duplicata

Registros que representam a mesma entidade real e que aparecem múltiplas vezes. Isso dá a falsa ideia ao modelo que esse evento é mais frequente do que ele realmente é, causando viéses e overfitting. Podem ser:

* Exact Duplicates
    * Podem ser causados por double click, ressubmissão de dados, falhas de rede que causam retransmissão
    * Verificar se a temporalidade dos eventos se tem importância
    * Tratamento simples. Apenas remove os demais registros, mantendo apenas a primeira ocorrência
* Near Duplicates
    * Representam a mesma entidade com pequenas diferenças
    * Exemplo: Variações de formato, como em (11)999999999 e 11999999999
    * Detectados por vários métodos
        * Similaridade de strings, como Jaro-Wrinkler e Levenshtein
        * Modelos de PLN para classificar matches
        * Token-based: Jaccard similarity, cosine similarity
    * Verificar a melhor estratégia de acordo com o tipo de diferenças
* Entity Duplicates
    * Registros distintos podem representar a mesma entidade, requerendo análise sofisticada
    * Exemplo
        * Customer Matching
            * João Silva, Rua A, 123, joao@email.com
            * J. Silva, R.A, 123, j.silva@company.com
        * Product Matching
            * "Iphone 12 64GB Preto"
            * "Apple iPhone 12 - 64GB - Black"
    * Podem ser detectados por deterministic ou probabilistic match, bem como ML similarity ou classification approach

### Feature Scaling

Todos os atributos precisam estar em uma escala para evitar que atributos com valores numéricos maiores tenham um impacto mais significativo que os demais. Devemos reescalar os atributos para que tenham escalas comparáveis.

Dataset: {Age 25-65, Income: 30.000-150.000, Experience: 0-40}

A distância Euclidiana sem Scaling 

* Z-score normalization
    * Subtrai a média e divide pelo desvio padrão, fazendo com que os dados tenham $\mu=0$ e $\omicron=1$
    * Preserva a forma do desvio normal
* Min-Max Scaling
    * Subtrai o valor do mínimo e divide pela diferença entre o máximo e o mínimo
    * Preserva as relações, mas não necessariamente ficam dentro do intervalo [0,1]
    * Consideramos sempre em relação o mínimo e máximo dos dados de treino pois são os únicos que temos acesso até o momento

### Encoding de variáveis categórias

Os algoritmos de ML apenas compreendem variáveis numéricas, mas às vezes temos dados categóricos, como texto, labels e categorias. Para garantir a compreensão do algoritmo, precismos converter categorias em representações numéricas sem perder informações importantes e sem introduzir relações artificiais.

A forma que tratamos depende dos tipos de variáveis categóricas:

* Nominais (sem ordem)
    * Não existe relação de grandeza entre as variáveis (cor)
    * Encoding adequado: One-hot, target encoding
* Ordinais (com ordem)
    * Existe a relação de grandeza entre as variáveis (tamanho, cargo)
    * Encoding adequado: ordinal encoding
* Alta cardinalidade (muitas categorias)
    * One-Hot cria muitas dimensões
    * Possíveis soluções: target encoding, hash encoding, embedding

1. Label Encoding
    * Cada categoria é mapeada para um número inteiro sequencial
    * Adequado para variáveis categóricas ordinais
    * Exemplo: {Pequeno, Médio, Grande} = {0, 1, 2}
2. Binary Encoding
    * Cada valor é convertido para uma representação binária
    * Tem uma boa dimensionalidade ($\log_{2}{k}$ colunas ao invés de $k$ no One-Hot)
    * Menos compreensível que o label encoding
    * Pode criar relações artificiais porque categorias com bits similares parecem mais próximas
3. One-Hot Encoding
    * Cria uma variável binária para cada categoria única
    * $k$ categorias = $k$ variáveis
    * Não impõe relação de ordem e ajuda bastante na interpretabilidade dos pesos porque existe uma clara correspondência entre categoria e variável
    * Muito esparso (cada linha tem soma 1) e consome muita memória
4. Dummy Encoding
    * Remove uma variável de referência, usando $k-1$ variáveis ao invés de $k$
5. [Target Encoding](https://medium.com/@heitornunesrosa/o-que-%C3%A9-target-encoding-e-como-aplic%C3%A1-lo-85996e1bf00d)
    * Substitui cada categoria pela estatística do target para cada categoria
    * Apenas uma variável independente do número de categorias, isto é, funciona com milhares de categorias
    * Bastante informativo porque cada categoria tem força baseada no target e nas estatísticas de interesse
6. Frequency Encoding
    * Substitui categoria por sua frequência relativa no dataset
    * Pode ser usado quando popularidade importa
    * Perde informações porque categorias diferentes podem ter a mesma frequência
7. Hash encoding
    * Usar função de hash para mapear categorias para dimensões fixas
    * Determinístico porque a mesma categoria sempre vai mapear pro mesmo hash
    * Podem ocorrer colisões

## Divisão dos dados em conjuntos de treino, validação e teste

Se usarmos os mesmos dados para treinar e avaliar um modelo, teremos uma ilusão de boa performance. Devemos treinar um modelo para que ele tenha uma boa capacidade de generalização e depois medir essa capacidade. Para isso, devemos dividir os dados em [conjuntos de treino, validação e teste](https://realpython.com/train-test-split-python-data/#training-validation-and-test-sets) antes das etapas de pré processamento.

Proporções típicas
* Datasets grandes (>100K observações)
    * Training: 60-70%
    * Validation: 15-20%
    * Test: 15-20%
* Datasets médios (10-100K observações)
    * Training: 70-80%
    * Validation: 10-25%
    * Test: 10-25%
* Datasets pequenos (<10K observações)
    * Training: 80%
    * Validation: Cross-Validation
    * Test: 20%

### Estratégia de divisão dos dados

* Random Split
    * A aleatoriedade pode não refletir bem o dataset
* Balanced Split
    * Pode não refletir distribuição real
* Stratified Split
    * Ambos os conjuntos mantém a distribuição original
* Time Series Split
    * NUNCA usar dados futuros para predizer eventos passados
    * Primeiros X% para treino, Y% para validação e Z% para teste
